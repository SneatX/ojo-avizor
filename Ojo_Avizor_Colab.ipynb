## 1. Subir el proyecto
Sube un **.zip con la carpeta del proyecto** (por ejemplo, la carpeta `ojo-avizor` comprimida) cuando aparezca el botón.

```python
from google.colab import files
import zipfile, os, glob

subido = files.upload()                      # elige el .zip con la carpeta del proyecto
nombre = next(iter(subido))
with zipfile.ZipFile(nombre) as z:
    z.extractall('/content/proyecto')
# Busca la carpeta que contiene app.py, se llame como se llame (ojo-avizor, ojo_avizor...).
app = sorted(glob.glob('/content/proyecto/**/app.py', recursive=True), key=len)[0]
os.chdir(os.path.dirname(app))
print('Proyecto listo en', os.getcwd())
print(sorted(os.listdir()))
```
